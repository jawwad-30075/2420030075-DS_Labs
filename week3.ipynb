```python
import numpy as np
from scipy.spatial import distance
pointA = np.array([2,4,6])
pointB = np.array([5,1,9])
euclidean_dist = distance.euclidean(pointA,pointB)
print("Euclidean Distance:",euclidean_dist)
similarity_euclidean = 1/(1+euclidean_dist)
print("Euclidean Distance:",similarity_euclidean)
```

```python
manhatten_dist  = distance.cityblock(pointA,pointB)
print("Manhatten Distance:",manhatten_dist)
similarity_manhatten = 1/(1+manhatten_dist)
print("Euclidean Distance:",similarity_manhatten)
```

```python

minkowski_dist_p3 = distance.minkowski(pointA, pointB, p=3)
print("Minkowski Distance:", minkowski_dist_p3)
similarity_minkowski = 1 / (1 + minkowski_dist_p3)
print("Minkowski Similarity:", similarity_minkowski)
```

```python
def hamming_distance(str1,str2):
    if len(str1) != len(str2):
        raise ValueError("Strings must be equal length")
    return sum(ch1 != ch2 for ch1, ch2 in zip(str1,str2))
s1 = "karolin"
s2 = "kathrin"
dist = hamming_distance(s1,s2)
print(f"Hamming Distance between '{s1}' and '{s2}':{dist}")
```

```python
def hamming_distance(str1,str2):
    if len(str1) != len(str2):
        raise ValueError("Strings must be equal length")
    return sum(ch1 != ch2 for ch1, ch2 in zip(str1,str2))
s1 = "karolin is full"
s2 = "kathrin is dull"
dist = hamming_distance(s1,s2)
print(f"Hamming Distance between '{s1}' and '{s2}':{dist}")
```

```python
def jaccard_index(str1, str2):
    set1, set2 = set(str1.split()), set(str2.split())
    inte = set1.intersection(set2)
    uni = set1.union(set2)
    return len(inte) / len(uni)

s1 = "data science is fun"
s2 = "science makes data useful"
print("Jaccard Index:", jaccard_index(s1, s2))
```

```python
from sklearn.feature_extraction.text import CountVectorizer
from sklearn.metrics.pairwise import cosine_similarity

s1 = "data science is fun"
s2 = "science makes data useful"

vectorizer = CountVectorizer().fit([s1, s2])
vectors = vectorizer.transform([s1, s2])

cos_sim = cosine_similarity(vectors[0], vectors[1])[0][0]
print("Cosine Similarity:", cos_sim)
```

```python
def lcs_length(x,y):
    m,n = len(x),len(y)
    dp = [[0]*(n+1) for _ in range(m+1)]
    for i in range(m):
        for j in range(n):
            if x[i] == y[j]:
                dp[i+1][j+1]=dp[i][j]
            else:
                dp[i+1][j+1] = max(dp[i][j+1],dp[i+1][j])
seq1 = "ABCDEF"
seq2 = "AEBDF"
lent = lcs_length(seq1,seq2)
print(f"Longest common Subsequence length between '{seq1}' and '{seq2}':{lent}")
                
```
